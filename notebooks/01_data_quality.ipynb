# FinSight AI — Notebook 01: Data Quality & Ingestion Validation

This notebook audits the raw synthetic transaction dataset, validates schema consistency, verifies mathematical balance reconciliation, and reports duplicate transaction handling.


```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent / 'src'))

import pandas as pd
import numpy as np
from finsight.ingestion.csv_loader import load_transactions_from_csv

tx_path = Path.cwd().parent / 'data' / 'synthetic' / 'transactions.csv'
print(f'Loading transaction data from: {tx_path}')
transactions, report = load_transactions_from_csv(tx_path)
print(f'Total records: {report.total_records}')
print(f'Valid records: {report.valid_records}')
print(f'Duplicates flagged: {report.duplicate_records}')
print(f'Reconciled: {report.reconciled}')
```

## 1. Missing Value and Data Completeness Audit

Let us inspect null values and feature availability across transaction fields.


```python
df_tx = pd.read_csv(tx_path)
print('Missing Values Count:')
print(df_tx.isnull().sum())

print('\nAccount Distribution:')
print(df_tx['account_id'].value_counts())
```
