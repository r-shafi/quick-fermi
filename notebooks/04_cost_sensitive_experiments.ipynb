# FinSight AI — Notebook 04: Cost-Sensitive Optimization & Asymmetric Error Curves

Demonstrates why standard 0.5 decision thresholds fail in financial shortfall intervention, and evaluates the cost-optimal decision threshold $\tau^*$ under $C_{FN}=5.0, C_{FP}=1.0$.


```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent / 'src'))

import numpy as np
from finsight.ml_core.cost_sensitive_loss import CostMatrix
from finsight.ml_core.threshold_selection import select_optimal_threshold

cm = CostMatrix(c_fn=5.0, c_fp=1.0)
print(f'Asymmetric Cost Formulation: C_FN={cm.c_fn}, C_FP={cm.c_fp}')
print(f'Analytical Bayes-Optimal Threshold: {cm.optimal_decision_threshold:.4f}')
```

## 1. Threshold Sensitivity Curve

Evaluate total cost across candidate thresholds from 0.01 to 0.99.


```python
y_val = np.array([0, 0, 0, 0, 0, 1, 1, 1, 0, 1])
val_probs = np.array([0.02, 0.05, 0.10, 0.12, 0.15, 0.22, 0.35, 0.65, 0.11, 0.70])

best_tau, min_cost, curve = select_optimal_threshold(val_probs, y_val, cm)
print(f'Optimal Threshold tau*: {best_tau:.4f}')
print(f'Minimum Financial Cost: {min_cost:.2f}')

cost_default, _ = cm.calculate_empirical_cost(y_val, (val_probs >= 0.5).astype(int))
print(f'Default Threshold 0.5 Cost: {cost_default:.2f} (Penalty Increase: +{cost_default - min_cost:.2f})')
```
