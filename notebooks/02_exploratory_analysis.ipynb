# FinSight AI — Notebook 02: Exploratory Financial Data Analysis

Explores transaction amounts, spending velocity, income periodicity, and liquidity trajectories across personal finance and SME accounts.


```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent / 'src'))

import pandas as pd
import numpy as np

tx_path = Path.cwd().parent / 'data' / 'synthetic' / 'transactions.csv'
gt_path = Path.cwd().parent / 'data' / 'synthetic' / 'ground_truth.csv'
df_tx = pd.read_csv(tx_path)
df_gt = pd.read_csv(gt_path)
print(f'Transactions: {len(df_tx)}, Ground Truth Days: {len(df_gt)}')
```

## 1. Spending by Category

Analyze essential obligations (Rent, Utilities, Payroll, Tax) vs discretionary spending.


```python
expenses = df_tx[df_tx['transaction_type'] == 'EXPENSE']
cat_summary = expenses.groupby('category')['amount'].agg(['count', 'sum', 'mean']).sort_values('sum', ascending=False)
print('Expense Summary by Category (BDT):')
print(cat_summary)
```

## 2. Ending Balance Trajectories Across Accounts


```python
acc_stats = df_gt.groupby('account_id')['ending_balance'].agg(['min', 'median', 'max'])
print('Ground-Truth Balance Extremes (BDT):')
print(acc_stats)
```
